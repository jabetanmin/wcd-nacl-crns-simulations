# Comparacion entre histogramas de Carga Total- PMT al 0.8 H/2

```python
import json
import matplotlib.pyplot as plt
import numpy as np
import math
```

```python
# Nombre del archivo de datos
archivo_datos = "Carga_Total_Agua_Pura_Neutrones_Termicos.txt"
# Lista para almacenar los datos
datos_NT_AP = []
# Abrir el archivo y leer los datos
with open(archivo_datos, 'r') as archivo:
    for linea in archivo:
        #Cada línea contiene un solo número en formato de texto
        valor = float(linea.strip())  # Convierte el texto a un número de punto flotante
        datos_NT_AP.append(valor)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import math
from matplotlib.lines import Line2D

# Definir número de bins y rango
numero_bins = 100
inicio, fin = 0, 100

# Función para calcular y graficar polígonos de frecuencia como línea continua
def plot_polygon(data, color, label):
    # Obtener histogram data
    counts, bins = np.histogram(data, bins=numero_bins, range=(inicio, fin))
    
    # Calcular el punto medio de cada bin
    bin_centers = (bins[:-1] + bins[1:]) / 2  
    
    # Graficar solo la línea continua (sin puntos)
    plt.plot(bin_centers, counts, linestyle='-', color=color, label=label, linewidth=2)

# Crear figura
plt.figure(figsize=(12, 8))

# Graficar los datos como polígonos de frecuencia (líneas continuas)
plot_polygon(datos_NT_AP, color='red', label='Agua pura')


# Agregar etiquetas y formato
plt.xlabel('ADC [u.a.]', fontsize=12)
plt.ylabel('Cuentas [u.a]', fontsize=12)
plt.yscale('log')  # Escala logarítmica
plt.xlim(0, 26)
plt.xticks(np.arange(0, 26, 2))
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.grid(axis="x", linestyle="--", alpha=0.7)
plt.tick_params(axis='x', which='both', labelsize=12)  # Tics mayores
plt.tick_params(axis='y', which='both', labelsize=12)  # Tics mayores
# Agregar leyenda
plt.legend(fontsize=24)

# Guardar y mostrar
plt.tight_layout()
plt.savefig('Histograma-Carga-Agua-pura-Termicos.jpg', format='jpg')
plt.savefig('Histograma-Carga-Agua-pura-Termicos.pdf', format='pdf')
plt.show()
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker


# ==========================================================
# CONFIGURACIÓN GENERAL
# ==========================================================

DIRECTORIO_SALIDA = "Graficas-numero-fotoelectrones"

os.makedirs(
    DIRECTORIO_SALIDA,
    exist_ok=True
)

ARCHIVO_PNG = os.path.join(
    DIRECTORIO_SALIDA,
    "Numero-fotoelectrones-neutrones-termicos.png"
)

ARCHIVO_PDF = os.path.join(
    DIRECTORIO_SALIDA,
    "Numero-fotoelectrones-neutrones-termicos.pdf"
)

ARCHIVO_JPG = os.path.join(
    DIRECTORIO_SALIDA,
    "Numero-fotoelectrones-neutrones-termicos.jpg"
)

ARCHIVO_DATOS = os.path.join(
    DIRECTORIO_SALIDA,
    "counts-number-photoelectrons.txt"
)


# ==========================================================
# CONFIGURACIÓN DEL HISTOGRAMA
# ==========================================================
#
# Como el número de fotoelectrones es una variable discreta,
# se emplean bins de ancho 1.
# ==========================================================

INICIO = 0
FIN = 200
ANCHO_BIN = 1

BORDES_BINS = np.arange(
    INICIO,
    FIN + ANCHO_BIN,
    ANCHO_BIN
)

CENTROS_BINS = (
    BORDES_BINS[:-1]
    + BORDES_BINS[1:]
) / 2.0


# ==========================================================
# OPCIONES DE LA FIGURA
# ==========================================================

MOSTRAR_INCERTIDUMBRE = False

LIMITE_X_MIN = 0
LIMITE_X_MAX = 160

DPI = 600

TITULO = (
    "Distribución del número de fotoelectrones "
    "producidos por neutrones térmicos"
)


# ==========================================================
# COLORES Y ESTILOS
# ==========================================================

CONFIGURACIONES = [
    {
        "datos": datos_NT_AP,
        "etiqueta": "Agua pura",
        "color": "#C62828",
        "linestyle": "-"
    },
    {
        "datos": datos_NT_A25NaCl,
        "etiqueta": r"Agua + 2.5 % de NaCl",
        "color": "#006400",
        "linestyle": "--"
    },
    {
        "datos": datos_NT_A5NaCl,
        "etiqueta": r"Agua + 5.0 % de NaCl",
        "color": "#6A1B9A",
        "linestyle": "-."
    },
    {
        "datos": datos_NT_A10NaCl,
        "etiqueta": r"Agua + 10.0 % de NaCl",
        "color": "#D97700",
        "linestyle": ":"
    }
]


# ==========================================================
# CONFIGURACIÓN TIPOGRÁFICA
# ==========================================================

plt.rcParams.update({
    "font.family": "serif",
    "font.serif": [
        "DejaVu Serif"
    ],
    "mathtext.fontset": "dejavuserif",

    "font.size": 17,
    "axes.titlesize": 22,
    "axes.labelsize": 24,
    "xtick.labelsize": 19,
    "ytick.labelsize": 19,
    "legend.fontsize": 17,

    "axes.linewidth": 1.1,

    # Mantiene las fuentes editables en PDF.
    "pdf.fonttype": 42,
    "ps.fonttype": 42
})


# ==========================================================
# FUNCIÓN PARA CALCULAR EL HISTOGRAMA
# ==========================================================

def obtener_histograma(datos):
    """
    Calcula las cuentas por bin y la incertidumbre estadística
    de Poisson sqrt(N).

    Parameters
    ----------
    datos : array-like
        Número de fotoelectrones por evento.

    Returns
    -------
    cuentas : ndarray
        Número de eventos por bin.

    incertidumbre : ndarray
        Incertidumbre estadística sqrt(N).
    """

    datos = np.asarray(
        datos,
        dtype=float
    )

    datos = datos[
        np.isfinite(datos)
    ]

    if datos.size == 0:
        raise ValueError(
            "Uno de los conjuntos de datos está vacío "
            "o no contiene valores numéricos válidos."
        )

    cuentas, _ = np.histogram(
        datos,
        bins=BORDES_BINS
    )

    cuentas = cuentas.astype(
        float
    )

    incertidumbre = np.sqrt(
        cuentas
    )

    return (
        cuentas,
        incertidumbre
    )


# ==========================================================
# CALCULAR LOS HISTOGRAMAS
# ==========================================================

resultados = []

for configuracion in CONFIGURACIONES:

    cuentas, incertidumbre = obtener_histograma(
        configuracion["datos"]
    )

    resultados.append({
        "etiqueta": configuracion["etiqueta"],
        "color": configuracion["color"],
        "linestyle": configuracion["linestyle"],
        "cuentas": cuentas,
        "incertidumbre": incertidumbre,
        "numero_eventos": int(
            np.sum(cuentas)
        )
    })


# ==========================================================
# CREAR LA FIGURA
# ==========================================================

fig, ax = plt.subplots(
    figsize=(13.2, 8.2),
    constrained_layout=True
)

fig.patch.set_facecolor(
    "white"
)

ax.set_facecolor(
    "white"
)


# ==========================================================
# GRAFICAR LAS DISTRIBUCIONES
# ==========================================================

for resultado in resultados:

    cuentas_visibles = np.where(
        resultado["cuentas"] > 0,
        resultado["cuentas"],
        np.nan
    )

    ax.stairs(
        cuentas_visibles,
        BORDES_BINS,
        color=resultado["color"],
        linestyle=resultado["linestyle"],
        linewidth=2.4,
        label=resultado["etiqueta"],
        zorder=5
    )

    if MOSTRAR_INCERTIDUMBRE:

        limite_inferior = np.where(
            resultado["cuentas"]
            - resultado["incertidumbre"]
            > 0,
            resultado["cuentas"]
            - resultado["incertidumbre"],
            np.nan
        )

        limite_superior = np.where(
            resultado["cuentas"] > 0,
            resultado["cuentas"]
            + resultado["incertidumbre"],
            np.nan
        )

        ax.fill_between(
            CENTROS_BINS,
            limite_inferior,
            limite_superior,
            step="mid",
            color=resultado["color"],
            alpha=0.10,
            linewidth=0,
            zorder=2
        )


# ==========================================================
# ESCALAS Y LÍMITES
# ==========================================================

ax.set_yscale(
    "log"
)

ax.set_xlim(
    LIMITE_X_MIN,
    LIMITE_X_MAX
)

# El mínimo se deja ligeramente por debajo de 1 para
# visualizar correctamente los bins con una sola cuenta.
ax.set_ylim(
    0.8,
    None
)


# ==========================================================
# TÍTULO Y ETIQUETAS
# ==========================================================

ax.set_title(
    TITULO,
    pad=15,
    weight="bold"
)

ax.set_xlabel(
    r"Número de fotoelectrones, $N_{\mathrm{pe}}$",
    labelpad=12
)

ax.set_ylabel(
    "Número de eventos por bin",
    labelpad=14
)


# ==========================================================
# CONFIGURACIÓN DE LOS TICKS
# ==========================================================

ax.set_xticks(
    np.arange(
        0,
        LIMITE_X_MAX + 1,
        20
    )
)

ax.xaxis.set_minor_locator(
    ticker.MultipleLocator(
        5
    )
)

ax.yaxis.set_major_locator(
    ticker.LogLocator(
        base=10,
        numticks=10
    )
)

ax.yaxis.set_minor_locator(
    ticker.LogLocator(
        base=10,
        subs=np.arange(2, 10) * 0.1,
        numticks=100
    )
)

ax.yaxis.set_major_formatter(
    ticker.LogFormatterMathtext(
        base=10
    )
)

ax.yaxis.set_minor_formatter(
    ticker.NullFormatter()
)

ax.tick_params(
    axis="both",
    which="major",
    direction="in",
    length=7,
    width=1.0,
    top=True,
    right=True
)

ax.tick_params(
    axis="both",
    which="minor",
    direction="in",
    length=3.5,
    width=0.65,
    top=True,
    right=True
)


# ==========================================================
# CUADRÍCULA
# ==========================================================

ax.grid(
    visible=True,
    which="major",
    axis="both",
    linestyle="--",
    linewidth=0.60,
    color="0.45",
    alpha=0.28
)

ax.grid(
    visible=True,
    which="minor",
    axis="both",
    linestyle=":",
    linewidth=0.35,
    color="0.60",
    alpha=0.12
)

ax.set_axisbelow(
    True
)


# ==========================================================
# LEYENDA
# ==========================================================

leyenda = ax.legend(
    loc="upper right",
    frameon=True,
    framealpha=0.96,
    facecolor="white",
    edgecolor="0.70",
    borderpad=0.65,
    labelspacing=0.55,
    handlelength=3.0
)

leyenda.get_frame().set_linewidth(
    0.8
)


# ==========================================================
# FORMATO DE LOS BORDES
# ==========================================================

for spine in ax.spines.values():

    spine.set_linewidth(
        1.1
    )

    spine.set_color(
        "0.20"
    )


# ==========================================================
# GUARDAR LA FIGURA
# ==========================================================

fig.savefig(
    ARCHIVO_PNG,
    format="png",
    dpi=DPI,
    bbox_inches="tight",
    pad_inches=0.05,
    facecolor="white"
)

fig.savefig(
    ARCHIVO_PDF,
    format="pdf",
    bbox_inches="tight",
    pad_inches=0.05,
    facecolor="white"
)

fig.savefig(
    ARCHIVO_JPG,
    format="jpg",
    dpi=DPI,
    quality=95,
    bbox_inches="tight",
    pad_inches=0.05,
    facecolor="white"
)

plt.show()

plt.close(
    fig
)


# ==========================================================
# GUARDAR LOS RESULTADOS NUMÉRICOS
# ==========================================================

with open(
    ARCHIVO_DATOS,
    "w",
    encoding="utf-8"
) as archivo:

    archivo.write(
        "# Distribución del número de fotoelectrones\n"
    )

    archivo.write(
        "# Bin de ancho 1 fotoelectrón\n"
    )

    archivo.write(
        "# Las incertidumbres corresponden a sqrt(N)\n"
    )

    archivo.write(
        "#\n"
    )

    archivo.write(
        "# E_min\t"
        "E_max\t"
        "E_centro\t"
        "Agua_pura\t"
        "Sigma_AP\t"
        "Agua_2p5_NaCl\t"
        "Sigma_2p5\t"
        "Agua_5p0_NaCl\t"
        "Sigma_5p0\t"
        "Agua_10p0_NaCl\t"
        "Sigma_10p0\n"
    )

    for i in range(
        len(CENTROS_BINS)
    ):

        archivo.write(
            f"{BORDES_BINS[i]:.1f}\t"
            f"{BORDES_BINS[i + 1]:.1f}\t"
            f"{CENTROS_BINS[i]:.2f}\t"
            f"{int(resultados[0]['cuentas'][i])}\t"
            f"{resultados[0]['incertidumbre'][i]:.6f}\t"
            f"{int(resultados[1]['cuentas'][i])}\t"
            f"{resultados[1]['incertidumbre'][i]:.6f}\t"
            f"{int(resultados[2]['cuentas'][i])}\t"
            f"{resultados[2]['incertidumbre'][i]:.6f}\t"
            f"{int(resultados[3]['cuentas'][i])}\t"
            f"{resultados[3]['incertidumbre'][i]:.6f}\n"
        )


# ==========================================================
# MOSTRAR INFORMACIÓN EN TERMINAL
# ==========================================================

print("\nFigura generada correctamente.")
print("------------------------------------------")

print(
    f"Directorio de salida:\n"
    f"{os.path.abspath(DIRECTORIO_SALIDA)}"
)

print(
    f"\nPNG:\n{ARCHIVO_PNG}"
)

print(
    f"\nPDF:\n{ARCHIVO_PDF}"
)

print(
    f"\nJPG:\n{ARCHIVO_JPG}"
)

print(
    f"\nDatos numéricos:\n{ARCHIVO_DATOS}"
)

print(
    f"\nNúmero de bins: "
    f"{len(CENTROS_BINS)}"
)

print(
    f"Ancho de bin: "
    f"{ANCHO_BIN} fotoelectrón"
)

for resultado in resultados:

    print(
        f"\n{resultado['etiqueta']}"
    )

    print(
        f"  Eventos dentro de 0–{FIN}: "
        f"{resultado['numero_eventos']:,}"
    )

    indice_maximo = int(
        np.argmax(
            resultado["cuentas"]
        )
    )

    print(
        f"  Bin modal: "
        f"{BORDES_BINS[indice_maximo]:.0f}–"
        f"{BORDES_BINS[indice_maximo + 1]:.0f}"
    )

    print(
        f"  Número máximo de cuentas: "
        f"{int(resultado['cuentas'][indice_maximo]):,}"
    )
```
