```python
import os
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker


# ==========================================================
# CONFIGURACIÓN GENERAL
# ==========================================================

OUTDIR = "Figura-espectro-electrones"
os.makedirs(OUTDIR, exist_ok=True)

ARCHIVO_PDF = os.path.join(
    OUTDIR,
    "Espectro-Electrones-sup-Bga.pdf"
)

ARCHIVO_PNG = os.path.join(
    OUTDIR,
    "Espectro-Electrones-sup-Bga.png"
)

# Intervalo energético mostrado
ENERGIA_MIN = 0.0
ENERGIA_MAX = 10.0

# Número de bins comunes para todos los histogramas.
# Puedes probar 5000, 10000 o 20000 según la estadística.
NUM_BINS = 10000

# Columna que contiene la energía.
# Índice 9 = décima columna del archivo.
COLUMNA_ENERGIA = 9


# ==========================================================
# ARCHIVOS, ETIQUETAS Y COLORES
# ==========================================================

# Se corrigió la correspondencia entre los archivos de
# 5 % y 10 % de NaCl.

configuraciones = [
    {
        "archivo": "Electrones-Agua-pura.txt",
        "etiqueta": "Agua pura",
        "color": "red",
        "zorder": 4
    },
    {
        "archivo": "Electrones-Agua-25NaCl.txt",
        "etiqueta": "Agua + 2.5 % de NaCl",
        "color": "green",
        "zorder": 5
    },
    {
        "archivo": "Electrones-Agua-5NaCl.txt",
        "etiqueta": "Agua + 5 % de NaCl",
        "color": "purple",
        "zorder": 6
    },
    {
        "archivo": "Electrones-Agua-10NaCl.txt",
        "etiqueta": "Agua + 10 % de NaCl",
        "color": "darkorange",
        "zorder": 7
    }
]


# ==========================================================
# CONFIGURACIÓN TIPOGRÁFICA
# ==========================================================

plt.rcParams.update({
    "font.family": "serif",
    "font.serif": ["DejaVu Serif"],
    "mathtext.fontset": "dejavuserif",

    "font.size": 18,
    "axes.labelsize": 24,
    "xtick.labelsize": 20,
    "ytick.labelsize": 20,
    "legend.fontsize": 18,

    "axes.linewidth": 1.0,

    # Mantener las fuentes editables en PDF
    "pdf.fonttype": 42,
    "ps.fonttype": 42
})


# ==========================================================
# FUNCIÓN PARA LEER LOS DATOS
# ==========================================================

def leer_energias(nombre_archivo, columna=9):
    """
    Lee la columna energética de un archivo de texto.

    Parameters
    ----------
    nombre_archivo : str
        Nombre o ruta del archivo.
    columna : int
        Índice de la columna energética.
        columna=9 corresponde a la décima columna.

    Returns
    -------
    numpy.ndarray
        Energías válidas en MeV.
    """

    if not os.path.exists(nombre_archivo):
        raise FileNotFoundError(
            f"No se encontró el archivo: {nombre_archivo}"
        )

    energias = []

    with open(
        nombre_archivo,
        "r",
        encoding="utf-8"
    ) as archivo:

        for numero_linea, linea in enumerate(
            archivo,
            start=1
        ):

            linea = linea.strip()

            # Ignorar líneas vacías y comentarios
            if not linea or linea.startswith("#"):
                continue

            columnas = linea.split()

            if len(columnas) <= columna:
                continue

            try:
                energia = float(columnas[columna])
            except ValueError:
                continue

            if np.isfinite(energia):
                energias.append(energia)

    if not energias:
        raise ValueError(
            f"No se encontraron energías válidas en "
            f"{nombre_archivo}."
        )

    return np.asarray(
        energias,
        dtype=float
    )


# ==========================================================
# LEER Y FILTRAR LOS DATOS
# ==========================================================

datos_por_archivo = {}

for configuracion in configuraciones:

    archivo = configuracion["archivo"]

    datos = leer_energias(
        archivo,
        columna=COLUMNA_ENERGIA
    )

    # Conservar únicamente el intervalo mostrado
    datos = datos[
        (datos >= ENERGIA_MIN) &
        (datos <= ENERGIA_MAX)
    ]

    datos_por_archivo[archivo] = datos

    print(
        f"{archivo}: {len(datos):,} electrones entre "
        f"{ENERGIA_MIN:.1f} y {ENERGIA_MAX:.1f} MeV"
    )


# ==========================================================
# BINS COMUNES PARA LOS CUATRO HISTOGRAMAS
# ==========================================================

bordes = np.linspace(
    ENERGIA_MIN,
    ENERGIA_MAX,
    NUM_BINS + 1
)

centros = 0.5 * (
    bordes[:-1] + bordes[1:]
)

ancho_bin = bordes[1] - bordes[0]

print(f"Número de bins: {NUM_BINS}")
print(f"Ancho de bin: {ancho_bin:.6e} MeV")


# ==========================================================
# CREAR FIGURA
# ==========================================================

fig, ax = plt.subplots(
    figsize=(13.5, 8.5),
    constrained_layout=True
)


# ==========================================================
# GRAFICAR LOS ESPECTROS
# ==========================================================

for configuracion in configuraciones:

    archivo = configuracion["archivo"]
    datos = datos_por_archivo[archivo]

    conteos, _ = np.histogram(
        datos,
        bins=bordes
    )

    # Los bins sin eventos no se dibujan en escala logarítmica.
    conteos_visibles = np.where(
        conteos > 0,
        conteos,
        np.nan
    )

    ax.plot(
        centros,
        conteos_visibles,
        linestyle="-",
        linewidth=1.35,
        color=configuracion["color"],
        label=configuracion["etiqueta"],
        zorder=configuracion["zorder"]
    )


# ==========================================================
# ESCALAS Y LÍMITES
# ==========================================================

ax.set_xlim(
    ENERGIA_MIN,
    ENERGIA_MAX
)

ax.set_yscale("log")

# El límite inferior debe ser positivo.
ax.set_ylim(
    0.8,
    None
)


# ==========================================================
# ETIQUETAS DE LOS EJES
# ==========================================================

ax.set_xlabel(
    r"Energía del electrón, $E_e$ [MeV]",
    labelpad=20
)

ax.set_ylabel(
    "Número de eventos",
    labelpad=20
)


# ==========================================================
# TICKS DEL EJE X
# ==========================================================

ax.xaxis.set_major_locator(
    ticker.MultipleLocator(1.0)
)

ax.xaxis.set_minor_locator(
    ticker.MultipleLocator(0.1)
)

ax.xaxis.set_major_formatter(
    ticker.FormatStrFormatter("%.0f")
)


# ==========================================================
# TICKS DEL EJE Y
# ==========================================================

ax.yaxis.set_major_locator(
    ticker.LogLocator(
        base=10,
        numticks=10
    )
)

ax.yaxis.set_minor_locator(
    ticker.LogLocator(
        base=10,
        subs=np.arange(2, 10) * 0.1,
        numticks=100
    )
)

ax.yaxis.set_major_formatter(
    ticker.LogFormatterMathtext(
        base=10
    )
)

ax.yaxis.set_minor_formatter(
    ticker.NullFormatter()
)


# ==========================================================
# APARIENCIA DE LOS TICKS
# ==========================================================

ax.tick_params(
    axis="both",
    which="major",
    direction="in",
    length=7,
    width=1.0,
    labelsize=20,
    top=True,
    right=True
)

ax.tick_params(
    axis="both",
    which="minor",
    direction="in",
    length=3.5,
    width=0.6,
    top=True,
    right=True
)


# ==========================================================
# CUADRÍCULA
# ==========================================================

ax.grid(
    visible=True,
    which="major",
    axis="both",
    linestyle="--",
    linewidth=0.55,
    color="0.55",
    alpha=0.35
)

ax.grid(
    visible=True,
    which="minor",
    axis="both",
    linestyle=":",
    linewidth=0.32,
    color="0.65",
    alpha=0.13
)

ax.set_axisbelow(True)


# ==========================================================
# IDENTIFICACIÓN DE LA SIMULACIÓN
# ==========================================================

ax.text(
    0.018,
    0.975,
    "Simulación Monte Carlo MEIGA",
    transform=ax.transAxes,
    fontsize=18,
    fontstyle="italic",
    color="0.30",
    ha="left",
    va="top",
    bbox=dict(
        facecolor="white",
        edgecolor="0.70",
        linewidth=0.7,
        boxstyle="round,pad=0.30",
        alpha=0.92
    ),
    zorder=15
)

# ==========================================================
# BORDE COMPTON DEL FOTÓN DE CAPTURA DEL HIDRÓGENO
# ==========================================================

E_COMPTON_H = 1.99

ax.axvline(
    E_COMPTON_H,
    color="0.35",
    linestyle=(0, (5, 4)),
    linewidth=1.2,
    alpha=0.9,
    zorder=2
)

ax.annotate(
    "Borde Compton\n"
    r"$E_{\gamma}=2.223$ MeV",
    xy=(E_COMPTON_H, 8e3),
    xytext=(2.55, 3e4),
    fontsize=16,
    ha="left",
    va="center",
    arrowprops=dict(
        arrowstyle="->",
        linewidth=1.1,
        color="0.30"
    ),
    bbox=dict(
        facecolor="white",
        edgecolor="0.65",
        alpha=0.92,
        boxstyle="round,pad=0.30"
    ),
    zorder=15
)

# ==========================================================
# LEYENDA
# ==========================================================

leyenda = ax.legend(
    loc="upper right",
    fontsize=18,
    frameon=True,
    framealpha=0.94,
    facecolor="white",
    edgecolor="0.65",
    borderpad=0.70,
    labelspacing=0.50,
    handlelength=2.5,
    handletextpad=0.75
)

leyenda.get_frame().set_linewidth(0.8)


# ==========================================================
# ESTILO DE LOS BORDES
# ==========================================================

for spine in ax.spines.values():

    spine.set_linewidth(1.0)
    spine.set_color("0.20")


# ==========================================================
# GUARDAR LA FIGURA
# ==========================================================

fig.savefig(
    ARCHIVO_PDF,
    format="pdf",
    bbox_inches="tight",
    pad_inches=0.06
)

fig.savefig(
    ARCHIVO_PNG,
    format="png",
    dpi=600,
    bbox_inches="tight",
    pad_inches=0.06
)


# ==========================================================
# MOSTRAR Y CERRAR
# ==========================================================

plt.show()
plt.close(fig)

print("Figura generada correctamente.")
print(f"PDF: {ARCHIVO_PDF}")
print(f"PNG: {ARCHIVO_PNG}")
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker


# ==========================================================
# CONFIGURACIÓN GENERAL
# ==========================================================

OUTDIR = "Figura-espectro-electrones"
os.makedirs(OUTDIR, exist_ok=True)

ARCHIVO_PDF = os.path.join(
    OUTDIR,
    "Espectro-Electrones-sup-Bga-Compton.pdf"
)

ARCHIVO_PNG = os.path.join(
    OUTDIR,
    "Espectro-Electrones-sup-Bga-Compton.png"
)

# Intervalo energético mostrado
ENERGIA_MIN = 0.0
ENERGIA_MAX = 10.0

# Número de bins comunes para todos los histogramas
NUM_BINS = 10000

# Índice 9 = décima columna del archivo
COLUMNA_ENERGIA = 9

# Masa en reposo del electrón
MASA_ELECTRON_MEV = 0.51099895


# ==========================================================
# ARCHIVOS, ETIQUETAS Y COLORES
# ==========================================================

configuraciones = [
    {
        "archivo": "Electrones-Agua-pura.txt",
        "etiqueta": "Agua pura",
        "color": "red",
        "zorder": 4
    },
    {
        "archivo": "Electrones-Agua-25NaCl.txt",
        "etiqueta": "Agua + 2.5 % de NaCl",
        "color": "green",
        "zorder": 5
    },
    {
        "archivo": "Electrones-Agua-5NaCl.txt",
        "etiqueta": "Agua + 5 % de NaCl",
        "color": "purple",
        "zorder": 6
    },
    {
        "archivo": "Electrones-Agua-10NaCl.txt",
        "etiqueta": "Agua + 10 % de NaCl",
        "color": "darkorange",
        "zorder": 7
    }
]


# ==========================================================
# LÍNEAS GAMMA Y BORDES COMPTON
# ==========================================================
#
# Cada elemento contiene:
#
# energia_gamma : energía de la línea gamma en MeV
# etiqueta       : texto que aparecerá en la figura
# color          : color de la línea vertical
# posicion_texto : coordenadas del texto (x, y)
#
# Las asociaciones nucleares deben contrastarse con el
# espectro gamma y la información de procesos de Geant4.
# ==========================================================

lineas_gamma = [
    {
        "energia_gamma": 2.223,
        "etiqueta": (
            r"$^{1}\mathrm{H}(n,\gamma){}^{2}\mathrm{H}$"
            "\n"
            r"$E_{\gamma}=2.223\ \mathrm{MeV}$"
        ),
        "color": "0.25",
        "posicion_texto": (2.18, 1.3e5),
        "ha": "left"
    },
    {
        "energia_gamma": 6.11,
        "etiqueta": (
            r"$E_{\gamma}=6.11\ \mathrm{MeV}$"
        ),
        "color": "0.40",
        "posicion_texto": (5.38, 2.6e4),
        "ha": "right"
    },
    {
        "energia_gamma": 6.92,
        "etiqueta": (
            r"$E_{\gamma}=6.92\ \mathrm{MeV}$"
        ),
        "color": "0.40",
        "posicion_texto": (6.42, 8.0e3),
        "ha": "right"
    },
    {
        "energia_gamma": 7.41,
        "etiqueta": (
            r"$E_{\gamma}=7.41\ \mathrm{MeV}$"
        ),
        "color": "0.40",
        "posicion_texto": (7.02, 2.2e3),
        "ha": "right"
    },
    {
        "energia_gamma": 8.58,
        "etiqueta": (
            r"$E_{\gamma}=8.58\ \mathrm{MeV}$"
        ),
        "color": "0.40",
        "posicion_texto": (8.12, 6.0e2),
        "ha": "right"
    }
]


# ==========================================================
# CONFIGURACIÓN TIPOGRÁFICA
# ==========================================================

plt.rcParams.update({
    "font.family": "serif",
    "font.serif": ["DejaVu Serif"],
    "mathtext.fontset": "dejavuserif",

    "font.size": 18,
    "axes.labelsize": 24,
    "xtick.labelsize": 20,
    "ytick.labelsize": 20,
    "legend.fontsize": 18,

    "axes.linewidth": 1.0,

    # Fuentes editables en PDF
    "pdf.fonttype": 42,
    "ps.fonttype": 42
})


# ==========================================================
# FUNCIÓN PARA LEER LOS DATOS
# ==========================================================

def leer_energias(nombre_archivo, columna=9):
    """
    Lee la columna energética de un archivo de texto.

    Parameters
    ----------
    nombre_archivo : str
        Nombre o ruta del archivo.

    columna : int
        Índice de la columna energética.
        columna=9 corresponde a la décima columna.

    Returns
    -------
    numpy.ndarray
        Energías válidas en MeV.
    """

    if not os.path.exists(nombre_archivo):
        raise FileNotFoundError(
            f"No se encontró el archivo: {nombre_archivo}"
        )

    energias = []

    with open(
        nombre_archivo,
        "r",
        encoding="utf-8"
    ) as archivo:

        for numero_linea, linea in enumerate(
            archivo,
            start=1
        ):

            linea = linea.strip()

            # Ignorar líneas vacías y comentarios
            if not linea or linea.startswith("#"):
                continue

            columnas = linea.split()

            if len(columnas) <= columna:
                continue

            try:
                energia = float(columnas[columna])

            except ValueError:
                continue

            if np.isfinite(energia):
                energias.append(energia)

    if not energias:
        raise ValueError(
            f"No se encontraron energías válidas en "
            f"{nombre_archivo}."
        )

    return np.asarray(
        energias,
        dtype=float
    )


# ==========================================================
# FUNCIÓN PARA CALCULAR EL BORDE COMPTON
# ==========================================================

def calcular_borde_compton(
    energia_gamma,
    masa_electron=MASA_ELECTRON_MEV
):
    """
    Calcula la energía cinética máxima transferida al electrón
    en una dispersión Compton.

    Parameters
    ----------
    energia_gamma : float
        Energía del fotón gamma en MeV.

    masa_electron : float
        Energía en reposo del electrón, m_e c^2, en MeV.

    Returns
    -------
    float
        Energía del borde Compton en MeV.
    """

    return energia_gamma * (
        1.0
        -
        1.0 / (
            1.0
            + 2.0 * energia_gamma / masa_electron
        )
    )


# ==========================================================
# CALCULAR Y MOSTRAR LOS BORDES COMPTON
# ==========================================================

for linea in lineas_gamma:

    linea["borde_compton"] = calcular_borde_compton(
        linea["energia_gamma"]
    )

    print(
        f"E_gamma = {linea['energia_gamma']:.3f} MeV "
        f"--> borde Compton = "
        f"{linea['borde_compton']:.3f} MeV"
    )


# ==========================================================
# LEER Y FILTRAR LOS DATOS
# ==========================================================

datos_por_archivo = {}

for configuracion in configuraciones:

    archivo = configuracion["archivo"]

    datos = leer_energias(
        archivo,
        columna=COLUMNA_ENERGIA
    )

    datos = datos[
        (datos >= ENERGIA_MIN)
        &
        (datos <= ENERGIA_MAX)
    ]

    datos_por_archivo[archivo] = datos

    print(
        f"{archivo}: {len(datos):,} electrones entre "
        f"{ENERGIA_MIN:.1f} y {ENERGIA_MAX:.1f} MeV"
    )


# ==========================================================
# BINS COMUNES PARA LOS CUATRO HISTOGRAMAS
# ==========================================================

bordes = np.linspace(
    ENERGIA_MIN,
    ENERGIA_MAX,
    NUM_BINS + 1
)

centros = 0.5 * (
    bordes[:-1]
    +
    bordes[1:]
)

ancho_bin = bordes[1] - bordes[0]

print(f"Número de bins: {NUM_BINS}")
print(f"Ancho de bin: {ancho_bin:.6e} MeV")


# ==========================================================
# CREAR FIGURA
# ==========================================================

fig, ax = plt.subplots(
    figsize=(13.5, 8.5),
    constrained_layout=True
)


# ==========================================================
# GRAFICAR LOS ESPECTROS
# ==========================================================

maximo_global = 0

for configuracion in configuraciones:

    archivo = configuracion["archivo"]
    datos = datos_por_archivo[archivo]

    conteos, _ = np.histogram(
        datos,
        bins=bordes
    )

    if conteos.size > 0:
        maximo_global = max(
            maximo_global,
            np.max(conteos)
        )

    # Los bins sin eventos no se dibujan en escala logarítmica
    conteos_visibles = np.where(
        conteos > 0,
        conteos,
        np.nan
    )

    ax.plot(
        centros,
        conteos_visibles,
        linestyle="-",
        linewidth=1.35,
        color=configuracion["color"],
        label=configuracion["etiqueta"],
        zorder=configuracion["zorder"]
    )


# ==========================================================
# ESCALAS Y LÍMITES
# ==========================================================

ax.set_xlim(
    ENERGIA_MIN,
    ENERGIA_MAX
)

ax.set_yscale("log")

ax.set_ylim(
    0.8,
    maximo_global * 1.8
)


# ==========================================================
# ETIQUETAS DE LOS EJES
# ==========================================================

ax.set_xlabel(
    r"Energía del electrón, $E_e$ [MeV]",
    labelpad=20
)

ax.set_ylabel(
    "Número de electrones",
    labelpad=20
)


# ==========================================================
# TICKS DEL EJE X
# ==========================================================

ax.xaxis.set_major_locator(
    ticker.MultipleLocator(1.0)
)

ax.xaxis.set_minor_locator(
    ticker.MultipleLocator(0.1)
)

ax.xaxis.set_major_formatter(
    ticker.FormatStrFormatter("%.0f")
)


# ==========================================================
# TICKS DEL EJE Y
# ==========================================================

ax.yaxis.set_major_locator(
    ticker.LogLocator(
        base=10,
        numticks=10
    )
)

ax.yaxis.set_minor_locator(
    ticker.LogLocator(
        base=10,
        subs=np.arange(2, 10) * 0.1,
        numticks=100
    )
)

ax.yaxis.set_major_formatter(
    ticker.LogFormatterMathtext(
        base=10
    )
)

ax.yaxis.set_minor_formatter(
    ticker.NullFormatter()
)


# ==========================================================
# APARIENCIA DE LOS TICKS
# ==========================================================

ax.tick_params(
    axis="both",
    which="major",
    direction="in",
    length=7,
    width=1.0,
    labelsize=20,
    top=True,
    right=True
)

ax.tick_params(
    axis="both",
    which="minor",
    direction="in",
    length=3.5,
    width=0.6,
    top=True,
    right=True
)


# ==========================================================
# CUADRÍCULA
# ==========================================================

ax.grid(
    visible=True,
    which="major",
    axis="both",
    linestyle="--",
    linewidth=0.55,
    color="0.55",
    alpha=0.35
)

ax.grid(
    visible=True,
    which="minor",
    axis="both",
    linestyle=":",
    linewidth=0.32,
    color="0.65",
    alpha=0.13
)

ax.set_axisbelow(True)


# ==========================================================
# IDENTIFICACIÓN DE LA SIMULACIÓN
# ==========================================================

ax.text(
    0.018,
    0.975,
    "Simulación Monte Carlo MEIGA",
    transform=ax.transAxes,
    fontsize=18,
    fontstyle="italic",
    color="0.30",
    ha="left",
    va="top",
    bbox=dict(
        facecolor="white",
        edgecolor="0.70",
        linewidth=0.7,
        boxstyle="round,pad=0.30",
        alpha=0.92
    ),
    zorder=20
)


# ==========================================================
# BORDES COMPTON
# ==========================================================

for indice, linea in enumerate(lineas_gamma):

    energia_gamma = linea["energia_gamma"]
    borde_compton = linea["borde_compton"]

    posicion_x_texto = linea["posicion_texto"][0]
    posicion_y_texto = linea["posicion_texto"][1]

    # Banda muy tenue alrededor del borde
    ancho_banda = 0.035

    ax.axvspan(
        borde_compton - ancho_banda,
        borde_compton + ancho_banda,
        facecolor=linea["color"],
        alpha=0.055,
        edgecolor="none",
        zorder=1
    )

    # Línea vertical del borde Compton
    ax.axvline(
        borde_compton,
        color=linea["color"],
        linestyle=(0, (5, 4)),
        linewidth=1.15,
        alpha=0.85,
        zorder=3
    )

    # Texto completo para la anotación
    texto_anotacion = (
        linea["etiqueta"]
        + "\n"
        + rf"$E_{{\mathrm{{C}}}}="
        + f"{borde_compton:.2f}"
        + r"\ \mathrm{MeV}$"
    )

    # Altura de la punta de la flecha
    # Se obtiene del máximo de las curvas alrededor del borde.
    mascara_local = (
        (centros >= borde_compton - 0.04)
        &
        (centros <= borde_compton + 0.04)
    )

    alturas_locales = []

    for configuracion in configuraciones:

        datos = datos_por_archivo[
            configuracion["archivo"]
        ]

        conteos_locales, _ = np.histogram(
            datos,
            bins=bordes
        )

        if np.any(mascara_local):

            valores_locales = conteos_locales[
                mascara_local
            ]

            if valores_locales.size > 0:

                alturas_locales.append(
                    np.max(valores_locales)
                )

    if alturas_locales:

        altura_flecha = max(
            max(alturas_locales),
            1.0
        )

    else:

        altura_flecha = 10.0

    ax.annotate(
        texto_anotacion,
        xy=(
            borde_compton,
            altura_flecha
        ),
        xytext=(
            posicion_x_texto,
            posicion_y_texto
        ),
        fontsize=13.5 if indice > 0 else 14.5,
        ha=linea["ha"],
        va="center",
        color="0.20",
        arrowprops=dict(
            arrowstyle="->",
            linewidth=1.0,
            color=linea["color"],
            shrinkA=3,
            shrinkB=3,
            connectionstyle="arc3,rad=0.0"
        ),
        bbox=dict(
            facecolor="white",
            edgecolor="0.70",
            linewidth=0.65,
            boxstyle="round,pad=0.25",
            alpha=0.90
        ),
        zorder=18
    )


# ==========================================================
# LEYENDA
# ==========================================================

leyenda = ax.legend(
    loc="upper right",
    fontsize=18,
    frameon=True,
    framealpha=0.94,
    facecolor="white",
    edgecolor="0.65",
    borderpad=0.70,
    labelspacing=0.50,
    handlelength=2.5,
    handletextpad=0.75
)

leyenda.get_frame().set_linewidth(0.8)


# ==========================================================
# ESTILO DE LOS BORDES
# ==========================================================

for spine in ax.spines.values():

    spine.set_linewidth(1.0)
    spine.set_color("0.20")


# ==========================================================
# GUARDAR LA FIGURA
# ==========================================================

fig.savefig(
    ARCHIVO_PDF,
    format="pdf",
    bbox_inches="tight",
    pad_inches=0.06
)

fig.savefig(
    ARCHIVO_PNG,
    format="png",
    dpi=600,
    bbox_inches="tight",
    pad_inches=0.06
)


# ==========================================================
# MOSTRAR Y CERRAR
# ==========================================================

plt.show()
plt.close(fig)

print("Figura generada correctamente.")
print(f"PDF: {ARCHIVO_PDF}")
print(f"PNG: {ARCHIVO_PNG}")
```
