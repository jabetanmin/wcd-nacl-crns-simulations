```python
import matplotlib.pyplot as plt
import numpy as np
import math

# Nombre de archivo de entrada y salida
archivo_entrada = '../Agua-pura/gamma-completo-AP-1000000meV.txt'
archivo_salida = 'gamma_AP_columna9_menor_10.5.txt'
archivo_histograma = 'histograma_datos_gamma_AP-2.txt'

# Abrir el archivo para lectura
with open(archivo_entrada, 'r') as archivo:
    data = []
    
    # Leer cada línea del archivo
    for linea in archivo:
        columnas = linea.split()
        valor_columna_9 = float(columnas[8])  # Columna 9 (índice 8)
        
        # Filtrar los menores a 10.5
        if valor_columna_9 < 10.5:
            data.append(valor_columna_9)

# Guardar los resultados en un archivo de texto
with open(archivo_salida, 'w') as f_out:
    for valor in data:
        f_out.write(f"{valor}\n")

print(f"Se guardaron {len(data)} valores en '{archivo_salida}'")

# ========================
# Parte de graficar histograma
# ========================

plt.figure(figsize=(12, 8))

# Histograma con 500000 bins
counts, bins, _ = plt.hist(data, bins=200, histtype='step', color='red', linewidth=1.7, log=True)

# Puntos en el centro de cada bin
bin_centers = 0.5 * (bins[:-1] + bins[1:])
plt.plot(bin_centers, counts, marker='.', linestyle='-', markersize=4, color='red')

# Guardar datos del histograma en archivo
with open(archivo_histograma, 'w') as f_hist:
    f_hist.write("BinCenter\tCounts\n")
    for x, y in zip(bin_centers, counts):
        f_hist.write(f"{x}\t{y}\n")

print(f"Se guardaron los datos del histograma en '{archivo_histograma}'")

# Ajustes de la gráfica
plt.tick_params(axis='y', which='both', direction='in', labelsize=20, right=True)
plt.title('')
plt.xlim(0, 10.01)
plt.xticks(np.arange(0, 10.01, 1), fontsize=24)
plt.yticks(fontsize=24)
plt.xlabel('Energy [MeV]', fontsize=24)
plt.ylabel('Counts [a.u.]', fontsize=24)
plt.yscale('log')
plt.grid(True)

# Guardar el gráfico
plt.savefig('Espectro-gammas-Agua-pura-1000000meV.pdf', format='pdf')
plt.savefig('Espectro-gammas-Agua-pura-1000000meV.jpg', format='jpg')

plt.show()
```
